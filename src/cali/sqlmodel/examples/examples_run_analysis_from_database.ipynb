# Run from Database Example

```python
from pathlib import Path

from sqlalchemy import create_engine

from cali.runner import CaliRunner
from cali.sqlmodel import AnalysisSettings
from cali.sqlmodel._model import Experiment
from cali.sqlmodel._visualize_experiment import print_cali_results
```

```python
database_path = "/Users/fdrgsp/Documents/git/cali/tests/test_data/evoked/results_1.cali"
engine = create_engine(f"sqlite:///{database_path}")
print_cali_results(engine, show_settings=False)
```

```python
# initialize CaliRunner
runner = CaliRunner()
```

```python
# get experiment from database
exp = Experiment.load_from_db(database_path)

# specify dataset path, detection settings ID, and analysis settings
dataset = "/Users/fdrgsp/Documents/git/cali/tests/test_data/evoked/evk.tensorstore.zarr"
analysis_settings = AnalysisSettings(dff_window=130)
detection_id = 2  # detection settings ID from the database (see printed results above)

# run analysis using new settings on existing detected ROIs (detection_id parameter)
runner.run(
    exp,
    dataset,
    detection_id,
    analysis_settings=analysis_settings,
    global_position_indices=[0],
    output_path=Path(database_path).parent,
    database_name="results_1.cali",
)
```

```python
print_cali_results(engine, show_settings=False)
```
